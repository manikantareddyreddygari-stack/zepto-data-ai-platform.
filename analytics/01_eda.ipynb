# Module 2: Analytics Pipeline — Part A: EDA, Cleaning, and Data Story
This notebook implements the analyst-to-data-scientist profiling workflow:
1. **Single Load**: Dataset loaded once via `sns.load_dataset('titanic')` and committed as `titanic.csv` offline fallback.
2. **Profiling**: `df.info()`, `df.describe()`, `df.shape`, and exact missing-value percentages.
3. **Threshold-Based Cleaning**: Under 5% dropped; 5%–30% imputed with median; >30% dropped with written justification.
4. **Univariate Analysis**: Histograms and box plots for `age` and `fare`; IQR outlier reporting; Skewness analysis of `fare`.
5. **Bivariate Analysis**: Boolean masking for survival rates by sex, pclass, and sex+pclass; 6x6 correlation matrix and heatmap.
6. **Multivariate Data Story**: 4 distinct charts with comprehensive written interpretations.
7. **Exploratory Standardization**: Z-score check on `age` and `fare`.

```python
import os
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from eda_pipeline import (
    load_and_profile_data,
    clean_data_by_threshold_rules,
    univariate_analysis,
    bivariate_analysis,
    multivariate_data_story,
    exploratory_standardization_check
)
%matplotlib inline
```

## Task 1: Dataset Profiling & Single Offline Fallback Load
The raw dataset is loaded once and saved as `titanic.csv` offline fallback.

```python
df_raw, missing_df = load_and_profile_data()
```

## Task 2: Threshold-Based Missing Value Handling
- `< 5%` missing: `embarked` (0.22%), `embark_town` (0.22%) $\to$ Drop affected rows.
- `5% - 30%` missing: `age` (19.87%) $\to$ Impute with median (28.00).
- `> 30%` missing: `deck` (77.22%) $\to$ Drop column due to severe sparsity.

```python
df_clean = clean_data_by_threshold_rules(df_raw, missing_df)
```

## Task 3: Univariate Analysis (Age & Fare)
- Outlier detection using the IQR rule: $[Q_1 - 1.5\times\text{IQR}, Q_3 + 1.5\times\text{IQR}]$
- Skewness analysis for fare comparing Mean, Median, and Mode.

```python
uni_results = univariate_analysis(df_clean)
```

## Task 4: Bivariate Analysis
- Survival rates broken down by sex, pclass, and sex & pclass combinations via boolean masking.
- 6x6 correlation matrix on numeric columns (`survived`, `pclass`, `age`, `sibsp`, `parch`, `fare`) excluding redundant boolean flags `adult_male` and `alone`.

```python
corr_df, top_2_corrs = bivariate_analysis(df_clean)
```

## Task 5: Multivariate Data Story
Four distinct charts establishing the core argument:
1. **Sex & Class Interaction**: Female passengers in 1st/2nd class achieved >92% survival.
2. **Age & Class Vulnerability**: Children in 1st/2nd class prioritized; elderly passengers in 3rd class had lowest survival.
3. **Fare & Economic Privilege**: Significantly higher fares correlated with survival within each cabin class.
4. **Family Size Dynamics**: Small families (size 2-4) had higher survival than solo travelers or large families (>4).

```python
multivariate_data_story(df_clean)
```

## Task 6: Exploratory Standardization Check
Computes $z = (x - \mu)/\sigma$ confirming $\mu \approx 0$ and $\sigma \approx 1$.

```python
exploratory_standardization_check(df_clean)
```
